```python
# Regular function
def x(t):
    x = 1
    for i in range(t):
        x = 2 * x
    return x

x(10)
```

```python
# Recursive function to do the same
def x(t):
    if t == 0:
        return 1
    else:
        return 2 * x(t-1)
# = 2*2*2*2...* x(0) = 2^10 * x(0) = 2^10 * 1
x(10)
```

```python
# Exercise 4.1 (my way)

def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)

factorial(10)
```

```python
# Exercise 4.1 (their solution)

def factorial(n):
    k = 1
    for i in range(n):
        k = k * (i + 1)
    return k

factorial(10)
```
