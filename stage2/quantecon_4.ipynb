```python
# Regular function
def x(t):
    x = 1
    for i in range(t):
        x = 2 * x
    return x

x(10)
```

```python
# Recursive function to do the same
def x(t):
    if t == 0:
        return 1
    else:
        return 2 * x(t-1)
# = 2*2*2*2...* x(0) = 2^10 * x(0) = 2^10 * 1
x(10)
```

```python
# Exercise 4.1 (their solution)

def factorial(n):
    k = 1
    for i in range(n):
        k = k * (i + 1)
    return k

factorial(10)
```

```python
# Exercise 4.2

import numpy as np

rng = np.random.default_rng()

def binomial_rv(n, p):
    count = 0
    for i in range(n):  
        U = rng.uniform()
        if U < p:
            count += 1
    return count

binomial_rv(1000, 0.65)
```

```python
# Exercise 4.4 - Regular way (non-recursive)

def fibonacci(t):
    
    x = [0, 1]

    for i in range(t):
        x.append(x[i] + x[i+1])
        print(x[i+1])
    print("This is the " + str(t) + "th number in the fibonacci sequence:")
    print(x[t])

fibonacci(13)
    
```

```python
# Exercise 4.4 - Recursive way

def x(t):
    if t == 0:
        return 0
    elif t == 1:
        return 1
    else:
        return x(t-1) + x(t-2)

print([x(i) for i in range(14)])

x(13)

```
