```python
example = np.zeros(10) # empty array of 10 items with 0s as placeholders

r = 0.025 # interest rate
T = 50 # end date
# Allocating memory at the start is more efficient than using a Python list and append,
# since the latter must repeatedly ask for storage space from the operating system
b = np.empty(T+1) # empty np array, used to store incoming data
b[0] = 10 # startig bal

for t in range(T):
    b[t+1] = (1 + r) * b[t]

plt.plot(b, label="bank balance")
plt.legend()
plt.show()
```

```python
# Exercise 3.1

import numpy as np
import matplotlib.pyplot as plt

T = 200
α = 0.9
x = np.empty(T+1)
x[0] = 0

rng = np.random.default_rng()
ϵ = rng.standard_normal(T)

for t in range(T):
    x[t+1] = α * x[t] + ϵ[t]

plt.plot(x)
plt.show()
```
