```python
a = ["a", "b", "c", "d", "e"]

print(a[2:5]) # a[m:n] returns n - m elements, starting at a[m]

print(a[-2:])

print(a[::3]) # [start:end:step] 

a[-2::-1] # walk backwards from the second to last element
```

```python
countries = ['Japan', 'Korea', 'China', 'US']
cities = ('Tokyo', 'Seoul', 'Beijing')
for country, city in zip(countries, cities): # zip pairs two sequences
    print(f'The capital of {country} is {city}')
```

```python
# if we want to get both the indeces (i) and values, a simple loop won't work
letter_list = ['a', 'b', 'c']
for i, letter in letter_list:
    print(f"letter_list[{i}] = '{letter}'")
```

```python
# if we want to get both the indeces (i) and values, we can use enumerate()
letter_list = ['a', 'b', 'c']
for i, letter in enumerate(letter_list):
    print(f"letter_list[{i}] = '{letter}'")
```
